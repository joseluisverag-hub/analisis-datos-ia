```python
hallazgos = {
    "supervivencia_general": "La mayoría de los pasajeros no sobrevivió; la tasa general de supervivencia fue menor al 50%.",
    "sexo": "Las mujeres tuvieron una tasa de supervivencia considerablemente más alta que los hombres.",
    "clase": "Los pasajeros de primera clase presentaron mejores tasas de supervivencia que los de segunda y tercera clase.",
    "sexo_clase": "Las mujeres de primera clase tuvieron la tasa más alta, mientras que los hombres de tercera clase presentaron la más baja.",
    "edad": "Los niños muestran mejores tasas de supervivencia que otros grupos etarios, mientras que los adultos mayores presentan las más bajas.",
    "familia": "Los grupos familiares pequeños o medianos muestran mejores tasas que quienes viajaban solos o en familias muy grandes."
}

hallazgos
```

```python
prompt_resumen = f"""
Actúa como analista de datos senior.

Necesito convertir los siguientes hallazgos de un análisis del Titanic en un resumen ejecutivo claro, profesional y breve.

Hallazgos:
- Supervivencia general: {hallazgos["supervivencia_general"]}
- Sexo: {hallazgos["sexo"]}
- Clase: {hallazgos["clase"]}
- Sexo + Clase: {hallazgos["sexo_clase"]}
- Edad: {hallazgos["edad"]}
- Tamaño de familia: {hallazgos["familia"]}

Instrucciones:
- Escribe en español.
- Usa tono ejecutivo y claro.
- No inventes datos nuevos.
- Resume en máximo 200 palabras.
- Entrega:
  1. Resumen ejecutivo
  2. 3 hallazgos clave
  3. 2 recomendaciones analíticas para un siguiente paso
"""

print(prompt_resumen)
```

```python
from pathlib import Path

Path("../outputs").mkdir(exist_ok=True)

with open("../outputs/prompt_resumen_titanic.txt", "w", encoding="utf-8") as f:
    f.write(prompt_resumen)

print("Prompt guardado en outputs/prompt_resumen_titanic.txt")
```

```python
resumen_ia = """
# Resumen Ejecutivo

El análisis del Titanic muestra que la supervivencia estuvo asociada principalmente a variables demográficas y socioeconómicas. La tasa general de supervivencia fue menor al 50%, confirmando la alta severidad del evento. El sexo aparece como uno de los factores más influyentes, con una clara ventaja para las mujeres frente a los hombres. La clase del pasajero también tuvo una relación importante con la supervivencia, observándose mejores resultados en primera clase y peores en tercera. La edad y el tamaño de la familia también muestran patrones relevantes: los niños tuvieron mejores tasas de supervivencia, mientras que los adultos mayores presentaron las más bajas; además, los grupos familiares pequeños o medianos obtuvieron mejores resultados que quienes viajaban solos o en familias muy grandes.

## Hallazgos clave
1. El sexo y la clase son los factores más asociados a la supervivencia.
2. Las mujeres de primera clase presentan la mayor tasa observada y los hombres de tercera clase la menor.
3. La edad y el tamaño de la familia aportan capacidad explicativa adicional.

## Recomendaciones analíticas
1. Construir un modelo predictivo usando variables como Sex, Pclass, Age y tamano_familia.
2. Evaluar interacciones entre variables para identificar perfiles de riesgo con mayor precisión.
"""

print(resumen_ia)
```

# IA aplicada al análisis

## Objetivo
Usar IA generativa para transformar hallazgos técnicos en un resumen ejecutivo claro y reutilizable.

## Enfoque
Se construyó un prompt estructurado a partir de los principales resultados del análisis exploratorio del Titanic.  
La salida buscada fue un resumen breve en lenguaje de negocio, acompañado de hallazgos clave y recomendaciones de siguiente paso.

## Valor agregado
Este flujo demuestra cómo la IA puede apoyar el trabajo del analista en tareas de:
- síntesis ejecutiva
- comunicación de insights
- documentación
- preparación de entregables para audiencias no técnicas

```python
def construir_prompt_resumen(hallazgos: dict) -> str:
    return f"""
Actúa como analista de datos senior.

Necesito convertir los siguientes hallazgos en un resumen ejecutivo claro, profesional y breve.

Hallazgos:
- Supervivencia general: {hallazgos["supervivencia_general"]}
- Sexo: {hallazgos["sexo"]}
- Clase: {hallazgos["clase"]}
- Sexo + Clase: {hallazgos["sexo_clase"]}
- Edad: {hallazgos["edad"]}
- Tamaño de familia: {hallazgos["familia"]}

Instrucciones:
- Escribe en español.
- Usa tono ejecutivo y claro.
- No inventes datos nuevos.
- Resume en máximo 200 palabras.
- Entrega:
  1. Resumen ejecutivo
  2. 3 hallazgos clave
  3. 2 recomendaciones analíticas
"""

prompt_generado = construir_prompt_resumen(hallazgos)
print(prompt_generado)
```
